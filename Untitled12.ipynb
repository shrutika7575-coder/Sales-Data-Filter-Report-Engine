```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 70)
print("             SALES DATA FILTER & REPORT ENGINE")
print("=" * 70)

np.random.seed(42)

stores = np.array([
    "Store 1", "Store 2", "Store 3", "Store 4", "Store 5",
    "Store 6", "Store 7", "Store 8", "Store 9", "Store 10"
])

months = np.arange(1, 13)

month_names = np.array([
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
])

sales = np.random.randint(20000, 100001, size=(10, 12))

sales_df = pd.DataFrame(sales, index=stores, columns=month_names)

print("\n1. ANNUAL SALES DATA")
print("-" * 70)
print(sales_df)

print("\nArray Shape:", sales.shape)
print("Number of Stores:", sales.shape[0])
print("Number of Months:", sales.shape[1])
print("Total Sales Entries:", sales.size)
print("Data Type:", sales.dtype)

store_totals = np.sum(sales, axis=1)
store_averages = np.mean(sales, axis=1)

print("\n\n2. STORE-WISE ANNUAL REPORT")
print("-" * 70)

store_report = pd.DataFrame({
    "Store": stores,
    "Annual Sales": store_totals,
    "Monthly Average": np.round(store_averages, 2)
})

print(store_report.to_string(index=False))

monthly_totals = np.sum(sales, axis=0)

print("\n\n3. MONTH-WISE SALES REPORT")
print("-" * 70)

for i in range(len(months)):
    print(f"{month_names[i]}: ₹{monthly_totals[i]:,}")

print("\n\n4. QUARTERLY SALES REPORT")
print("-" * 70)

quarters = {
    "Q1 (Jan-Mar)": sales[:, 0:3],
    "Q2 (Apr-Jun)": sales[:, 3:6],
    "Q3 (Jul-Sep)": sales[:, 6:9],
    "Q4 (Oct-Dec)": sales[:, 9:12]
}

quarter_totals = {}

for quarter_name, quarter_data in quarters.items():
    total = np.sum(quarter_data)
    quarter_totals[quarter_name] = total
    print(f"{quarter_name}: ₹{total:,}")

best_quarter = max(quarter_totals, key=quarter_totals.get)

print("\nBest Performing Quarter:", best_quarter)

print("\n\n5. SALES ABOVE TARGET")
print("-" * 70)

target = 70000
high_sales = sales > target

print("Sales Target: ₹", target)
print("Number of Sales Entries Above Target:", np.sum(high_sales))

for i, j in np.argwhere(high_sales):
    print(f"{stores[i]} - {month_names[j]}: ₹{sales[i, j]:,}")

print("\n\n6. STORES BELOW AVERAGE")
print("-" * 70)

overall_average = np.mean(sales)
below_average = store_averages < overall_average

print("Overall Monthly Sales Average: ₹", round(overall_average, 2))

for i in np.where(below_average)[0]:
    print(f"{stores[i]}: ₹{store_averages[i]:,.2f}")

print("\n\n7. COMBINED BOOLEAN FILTERING")
print("-" * 70)

high_performing_stores = (store_averages > overall_average) & (store_totals > np.mean(store_totals))

print("Stores performing above the average:")

for store in stores[high_performing_stores]:
    print(store)

print("\nMonths with total sales above ₹700,000:")

high_months = monthly_totals > 700000

for month in month_names[high_months]:
    print(month)

print("\n\n8. FANCY INDEXING")
print("-" * 70)

selected_indices = np.array([1, 4, 7])

print("Sales of selected stores (Store 2, Store 5, Store 8):")
print(sales[selected_indices, :])

top_month_indices = np.argsort(monthly_totals)[-3:][::-1]

print("\nTop 3 Sales Months:")

for i in top_month_indices:
    print(f"{month_names[i]}: ₹{monthly_totals[i]:,}")

print("\n\n9. GROWTH TARGET ANALYSIS")
print("-" * 70)

growth_targets = np.linspace(500000, 800000, 12)

growth_report = pd.DataFrame({
    "Month": month_names,
    "Actual Sales": monthly_totals,
    "Growth Target": growth_targets.astype(int),
    "Target Achieved": monthly_totals >= growth_targets
})

print(growth_report.to_string(index=False))

print("\n\n10. STORE-WISE SALES COMPARISON")
print("-" * 70)

best_store_index = np.argmax(store_totals)
worst_store_index = np.argmin(store_totals)

print("Top Store:", stores[best_store_index])
print("Annual Sales: ₹", store_totals[best_store_index])

print("Lowest Performing Store:", stores[worst_store_index])
print("Annual Sales: ₹", store_totals[worst_store_index])

print("\n\n11. MONTHLY SALES GRAPH")
print("-" * 70)

plt.figure(figsize=(12, 5))
plt.plot(month_names, monthly_totals, marker="o")
plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Total Sales (₹)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print("\n\n12. QUARTERLY SALES GRAPH")
print("-" * 70)

plt.figure(figsize=(9, 5))
plt.bar(list(quarter_totals.keys()), list(quarter_totals.values()))
plt.title("Quarterly Sales Comparison")
plt.xlabel("Quarter")
plt.ylabel("Total Sales (₹)")
plt.xticks(rotation=15)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print("\n\n13. STORE-WISE ANNUAL SALES GRAPH")
print("-" * 70)

plt.figure(figsize=(12, 5))
plt.bar(stores, store_totals)
plt.title("Annual Sales by Store")
plt.xlabel("Stores")
plt.ylabel("Annual Sales (₹)")
plt.xticks(rotation=30)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

print("\n\n" + "=" * 70)
print("                  FINAL PROJECT SUMMARY")
print("=" * 70)

print("Total Stores:", len(stores))
print("Total Months:", len(months))
print("Total Annual Sales: ₹", np.sum(sales))
print("Average Monthly Sales Across All Stores: ₹", round(overall_average, 2))
print("Best Performing Store:", stores[best_store_index])
print("Best Performing Month:", month_names[np.argmax(monthly_totals)])
print("Best Performing Quarter:", best_quarter)
print("Sales Entries Above Target:", np.sum(high_sales))
print("\nProject executed successfully!")
print("=" * 70)

```
